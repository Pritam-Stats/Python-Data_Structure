### **3. Write a code to check if a given string is a palindrome or not**

```python
def is_palindrome(s):
    return s == s[::-1] #if the rev is same

# Example
print(is_palindrome("radar"))
```

### **4. Write a code to check if two given strings are anagrams of each other**

```python
def are_anagrams(s1, s2):
    """
    An anagram is a word or phrase formed by rearranging the letters of a different word or phrase, typically using all the original letters exactly once. For example, the word "secure" is an anagram of "rescue".
    """
    return sorted(s1) == sorted(s2)

# Example
print(are_anagrams("listen", "silent")) #t
print(are_anagrams("secure", "rescue")) #T
```

### **5. Write a code to find all occurrences of a given substring within another string**

### **6. Write a code to perform basic string compression using the counts of repeated characters**

```python
def compress_string(s):
    compressed = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            compressed.append(s[i - 1] + str(count))
            count = 1
    compressed.append(s[-1] + str(count))
    return ''.join(compressed)  #string method

# Example
print(compress_string("aabcccccaaa"))
```

### **12. Implement a code to find the second largest number in a given list of integers**

```python
def second_largest(nums):
    unique_nums = list(set(nums))
    unique_nums.sort()
    return unique_nums[-2] if len(unique_nums) >= 2 else None
    #if the list len is 1 then there can't be a second largest

# Example
print(second_largest([10, 20, 4, 45, 99]))
```

### **14. Reverse a list in-place without using built-in reverse functions**

```python
def reverse_list_inplace(lst):
    left, right = 0, len(lst) - 1
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]
        left += 1
        right -= 1
    return lst

# Example
print(reverse_list_inplace([1, 2, 3, 4, 5]))
```

### **15. Remove duplicates from a list while preserving order**

```python
def remove_duplicates(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

# Example
print(remove_duplicates([1, 2, 2, 3, 1, 4]))
```

### **16. Check if a list is sorted (ascending or descending)**

### **17. Merge two sorted lists into one sorted list**

```python
def merge_sorted_lists(a, b):
    result = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    result.extend(a[i:])
    result.extend(b[j:])
    return result

# Example
print(merge_sorted_lists([1, 3, 5], [2, 4, 6]))
```

### **20. Shuffle a list randomly without using built-in shuffle**

```python
import random

def manual_shuffle(lst):
    result = lst[:]
    for i in range(len(result) - 1, 0, -1):
        j = random.randint(0, i)
        result[i], result[j] = result[j], result[i]
    return result

# Example
print(manual_shuffle([1, 2, 3, 4, 5]))
```

### **31. Frequency dictionary of words in list**

```python
def word_frequencies(words):
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

# Example
print(word_frequencies(['apple', 'banana', 'apple', 'orange']))
```

### **32. Merge two dictionaries, summing values for common keys**

```python
def merge_dicts(d1, d2):
    result = d1.copy()
    for k, v in d2.items():
        result[k] = result.get(k, 0) + v
    return result

# Example
print(merge_dicts({'a': 1, 'b': 2}, {'b': 3, 'c': 4}))
```

### **33. Access value in nested dictionary using list of keys**

```python
def access_nested(d, keys):
    for key in keys:
        d = d.get(key)
        if d is None:
            return None
    return d

# Example
print(access_nested({'a': {'b': {'c': 42}}}, ['a', 'b', 'c']))
```

### **34. Return dictionary sorted by values (ascending or descending)**

```python
def sort_dict_by_value(d, reverse=False):
    return dict(sorted(d.items(), key=lambda item: item[1], reverse=reverse))

# Example
print(sort_dict_by_value({'a': 3, 'b': 1, 'c': 2}))
```

### **35. Invert a dictionary, grouping keys by shared values**

```python
def invert_dict(d):
    inv = {}
    for k, v in d.items():
        inv.setdefault(v, []).append(k)
    return inv

# Example
print(invert_dict({'a': 1, 'b': 2, 'c': 1}))
```
